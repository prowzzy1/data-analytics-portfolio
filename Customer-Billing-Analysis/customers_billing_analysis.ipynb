```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='account_status', data=df)
plt.title("Account Status Distribution")
plt.show()
```

```python
sns.boxplot(x='account_status', y='amount_due', data=df)
plt.title("Amount Due by Account Status")
plt.show()
```

```python
sns.countplot(x='issue_type', data=df)
plt.xticks(rotation=45)
plt.title("Common Customer Issues")
plt.show()
```

```python
sns.countplot(x='billing_error', data=df)
plt.title("Billing Errors Distribution")
plt.show()
```

## Key Insights

- Suspended accounts tend to have higher outstanding balances.
- Billing errors are a common issue and may contribute to customer dissatisfaction.
- The most frequent customer issues are late payments and overcharges.
- Customers with more complex issues have longer resolution times.

## Business Recommendations

- Improve billing accuracy to reduce errors.
- Implement early alerts for high balances to prevent account suspension.
- Focus on resolving common issues faster to improve customer experience.
